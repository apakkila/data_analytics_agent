```python
import pandas as pd
import os
```

```python
pwd = os.getcwd()
csv_path = os.path.join(pwd, '..', 'data', 'ncr_ride_bookings.csv')
print(csv_path)
df = pd.read_csv(csv_path, encoding='latin1', on_bad_lines='skip')
```

```python
print(df.columns)
```

```python
print(df.head())
```

```python
df['Timestamp'] = df['Date'] + ' ' + df['Time']
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
```

```python
# variables that need to be converted from NaN to 0

print(f'Unique values for driver ratings {df["Driver Ratings"].unique()}')
print(f'Unique values for customer ratings {df["Customer Rating"].unique()}')
print(f'Unique values for booking status {df["Booking Status"].unique()}')
print(f'Unique values for payment mode {df["Payment Method"].unique()}')


# Print min and max values for Avg VTAT, Avg CTAT, Booking Value, Ride Distance
print(f'Avg VTAT: min={df["Avg VTAT"].min()}, max={df["Avg VTAT"].max()}')
print(f'Avg CTAT: min={df["Avg CTAT"].min()}, max={df["Avg CTAT"].max()}')
print(f'Booking Value: min={df["Booking Value"].min()}, max={df["Booking Value"].max()}')
print(f'Ride Distance: min={df["Ride Distance"].min()}, max={df["Ride Distance"].max()}')

# 
```

```python
# Convert nan values in columns 'Incomplete Rides', 'Cancelled Rides by Customer', 'Cancelled Rides by Driver', 'Driver Ratings', 'Customer Rating', 'Avg VTAT', 'Avg CTAT', 'Booking Value', 'Ride Distance' columns to 0

df['Pickup Location'] = df['Pickup Location'].fillna('Unknown')
df['Drop Location'] = df['Drop Location'].fillna('Unknown')
df['Driver Ratings'] = df['Driver Ratings'].fillna(0)
df['Customer Rating'] = df['Customer Rating'].fillna(0)
df['Avg VTAT'] = df['Avg VTAT'].fillna(0)
df['Avg CTAT'] = df['Avg CTAT'].fillna(0)
df['Booking Value'] = df['Booking Value'].fillna(0)
df['Ride Distance'] = df['Ride Distance'].fillna(0)
df['Reason for cancelling by Customer'] = df['Reason for cancelling by Customer'].fillna('No Reason Provided')
df['Driver Cancellation Reason'] = df['Driver Cancellation Reason'].fillna('No Reason Provided')
df['Incomplete Rides Reason'] = df['Incomplete Rides Reason'].fillna('No Reason Provided')
df['Payment Method'] = df['Payment Method'].fillna('Unknown')
```

```python
# Ensure that the values in each column are of the correct data type

# String columns
df['Pickup Location'] = df['Pickup Location'].astype(str)
df['Drop Location'] = df['Drop Location'].astype(str)
df['Reason for cancelling by Customer'] = df['Reason for cancelling by Customer'].astype(str)
df['Driver Cancellation Reason'] = df['Driver Cancellation Reason'].astype(str)
df['Incomplete Rides Reason'] = df['Incomplete Rides Reason'].astype(str)
df['Payment Method'] = df['Payment Method'].astype(str)
df['Customer ID'] = df['Customer ID'].astype(str)
df['Booking Status'] = df['Booking Status'].astype(str)

# Numeric columns
df['Booking Value'] = pd.to_numeric(df['Booking Value'], errors='coerce')
df['Ride Distance'] = pd.to_numeric(df['Ride Distance'], errors='coerce')
df['Avg VTAT'] = pd.to_numeric(df['Avg VTAT'], errors='coerce')
df['Avg CTAT'] = pd.to_numeric(df['Avg CTAT'], errors='coerce')

# Ratings (Can be float or int depending on data source, using float for generality)
df['Driver Ratings'] = pd.to_numeric(df['Driver Ratings'], errors='coerce')
df['Customer Rating'] = pd.to_numeric(df['Customer Rating'], errors='coerce')
```

```python
# VTAT = waiting time, CTAT = ride time

columns_to_keep = ['Timestamp', 'Booking Status', 'Customer ID', 'Booking ID',
       'Pickup Location', 'Drop Location', 'Avg VTAT',
       'Avg CTAT',
       'Reason for cancelling by Customer',
       'Driver Cancellation Reason',
       'Incomplete Rides Reason', 'Booking Value', 'Ride Distance',
       'Driver Ratings', 'Customer Rating', 'Payment Method']

df_cleaned = df[columns_to_keep]
```

```python
import pandas as pd
# Assuming 'df' is the DataFrame resulting from your initial cleaning script

def rename_columns_to_snake_case(df):
    """
    Renames the columns of the DataFrame to snake_case using a direct mapping.
    
    IMPORTANT: This assumes the input DataFrame 'df' still has the original
    column names as its starting point.
    """
    
    # 1. Define the complete mapping from RAW NAME to FINAL SNAKE_CASE NAME
    column_name_mapping = {
        # Time and Status
        'Timestamp': 'ride_timestamp',
        'Booking Status': 'booking_status',
        'Customer ID': 'customer_id',
        'Booking ID': 'booking_id',
        
        # Location and Timing
        'Pickup Location': 'pickup_location',
        'Drop Location': 'drop_location',
        'Avg VTAT': 'avg_waiting_time',  # Using avg_vitat as per your SQL schema definition
        'Avg CTAT': 'avg_ride_time',   # Using avg_ctat as per your SQL schema definition
        
        # Reasons
        'Reason for cancelling by Customer': 'customer_cancellation_reason',
        'Driver Cancellation Reason': 'driver_cancellation_reason',
        'Incomplete Rides Reason': 'incomplete_rides_reason',
        
        # Financial and Ratings
        'Booking Value': 'booking_value',
        'Ride Distance': 'ride_distance',
        'Driver Ratings': 'driver_ratings',
        'Customer Rating': 'customer_rating',
        'Payment Method': 'payment_method'
    }

    # 2. Rename the columns using the direct mapping
    # This automatically filters the DataFrame to the columns in the mapping.
    df_cleaned = df.rename(columns=column_name_mapping)
    
    print("--- Column Rename Complete ---")
    print("New DataFrame Columns (Snake Case):")
    print(df_cleaned.columns.tolist())
    
    return df_cleaned

# You need to ensure the variable 'df' being passed here is the one right AFTER
# reading the CSV (or before any snake_casing).
df_cleaned = rename_columns_to_snake_case(df_cleaned)
```

```python
print(df_cleaned.isnull().sum().sort_values(ascending=False))
```

```python
# In Pandas (before saving the CSV)
# Check for duplicates:
print(df_cleaned.duplicated(subset=['booking_id']).sum())

# Keep only the first instance of each booking_id (or the most complete row):
df_deduplicated = df_cleaned.drop_duplicates(subset=['booking_id'], keep='first')
print(df_deduplicated.duplicated(subset=['booking_id']).sum())
```

```python
# Print unique values for each column in df_deduplicated
for column in df_deduplicated.columns:
    print(f"Unique values for {column}:")
    print(df_deduplicated[column].unique())
    print("\n")
```

```python


# Save cleaned data to new CSV file
cleaned_csv_path = os.path.join(pwd, '..', 'data', 'ncr_ride_bookings_cleaned.csv')
df_deduplicated.to_csv(cleaned_csv_path, index=False)
```
